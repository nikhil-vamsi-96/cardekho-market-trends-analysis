# CarDekho Market Trends Analysis

**Goal:** Analyze used-vehicle listings to understand price trends, depreciation, vehicle age, mileage, fuel type, transmission, seller type, and ownership.

**Dataset:** CarDekho used-vehicle data.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/car_data.csv")
df.head()
```

## 1. Basic Dataset Overview


```python
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("\nColumns:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nManufacturing year range:", df["Year"].min(), "to", df["Year"].max())
print("Lowest selling price:", df["Selling_Price"].min())
print("Highest selling price:", df["Selling_Price"].max())
print("Most listed vehicle:")
print(df["Car_Name"].value_counts().head(1))
```

## 2. Price and Depreciation Analysis


```python
df["Depreciation"] = df["Present_Price"] - df["Selling_Price"]
df["Vehicle_Age"] = 2026 - df["Year"]

print("Highest depreciation:")
print(df.loc[df["Depreciation"].idxmax(), ["Car_Name","Year","Present_Price","Selling_Price","Depreciation"]])

print("\nLowest depreciation:")
print(df.loc[df["Depreciation"].idxmin(), ["Car_Name","Year","Present_Price","Selling_Price","Depreciation"]])
```

```python
yearly_price = df.groupby("Year", as_index=False)["Selling_Price"].mean()

plt.figure(figsize=(9,5))
plt.plot(yearly_price["Year"], yearly_price["Selling_Price"], marker="o")
plt.title("Average Selling Price by Manufacturing Year")
plt.xlabel("Manufacturing Year")
plt.ylabel("Average Selling Price (Lakhs)")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.scatterplot(data=df, x="Kms_Driven", y="Selling_Price")
plt.title("Selling Price vs Kilometres Driven")
plt.xlabel("Kilometres Driven")
plt.ylabel("Selling Price (Lakhs)")
plt.tight_layout()
plt.show()
```

```python
age_dep = df.groupby("Vehicle_Age", as_index=False)["Depreciation"].mean()

plt.figure(figsize=(9,5))
plt.plot(age_dep["Vehicle_Age"], age_dep["Depreciation"], marker="o")
plt.title("Average Price Depreciation vs Vehicle Age")
plt.xlabel("Vehicle Age (Years)")
plt.ylabel("Average Depreciation (Lakhs)")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

## 3. Market Segmentation


```python
print("Fuel type counts:")
print(df["Fuel_Type"].value_counts())

print("\nTransmission counts:")
print(df["Transmission"].value_counts())

print("\nSeller type counts:")
print(df["Seller_Type"].value_counts())

print("\nPrevious owner counts:")
print(df["Owner"].value_counts().sort_index())

print("\nCNG vehicles:", (df["Fuel_Type"] == "CNG").sum())
print("Automatic vehicles:", (df["Transmission"] == "Automatic").sum())
print("Individual sellers:", (df["Seller_Type"] == "Individual").sum())
print("Single-owner vehicles (Owner=0):", (df["Owner"] == 0).sum())
```

## 4. Cars vs Two-Wheelers


```python
bike_patterns = (
    "activa|bajaj|hero |honda cb|honda cbr|honda karizma|honda dream|"
    "hyosung|ktm|mahindra mojo|royal enfield|suzuki access|tvs apache|"
    "tvs jupyter|tvs sport|tvs wego|um renegade|yamaha fz|yamaha fazer"
)

is_bike = df["Car_Name"].astype(str).str.lower().str.contains(bike_patterns, regex=True, na=False)
bikes = df[is_bike].copy()
cars = df[~is_bike].copy()

print("Two-wheelers:", len(bikes))
print("Cars:", len(cars))

if len(bikes):
    print("\nOldest bike:")
    print(bikes.loc[bikes["Year"].idxmin(), ["Car_Name","Year","Selling_Price"]])
    print("\nNewest bike:")
    print(bikes.loc[bikes["Year"].idxmax(), ["Car_Name","Year","Selling_Price"]])
    print("\nMost listed bike:")
    print(bikes["Car_Name"].value_counts().head(1))

print("\nOldest car:")
print(cars.loc[cars["Year"].idxmin(), ["Car_Name","Year","Selling_Price"]])
print("\nNewest car:")
print(cars.loc[cars["Year"].idxmax(), ["Car_Name","Year","Selling_Price"]])
print("\nMost listed car:")
print(cars["Car_Name"].value_counts().head(1))
```

## Conclusion

The analysis shows clear variation in resale prices across vehicle age, mileage, fuel type, transmission, seller type, and ownership history. These insights can help buyers compare vehicles, sellers estimate resale value, and dealers understand used-vehicle market patterns.
